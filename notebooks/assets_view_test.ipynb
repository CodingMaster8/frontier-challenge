```python
conn = duckdb.connect('../data/br_funds.db')

# Get count of each issuer type
df = conn.execute("""
    SELECT 
        UNNEST(string_split(issuer_type, ', ')) as issuer_type,
        COUNT(*) as count
    FROM fund_holdings_detail_view
    WHERE issuer_type IS NOT NULL
    GROUP BY issuer_type
    ORDER BY count DESC
""").fetchdf()

conn.close()

df
```

```python
conn = duckdb.connect('../data/br_funds.db')

# Example: Find similar issuer names using Levenshtein distance
df = conn.execute("""
    SELECT 
        cnpj,
        issuer_name,
        levenshtein(issuer_name, 'PETROBRAS') as distance
    FROM fund_holdings_detail_view
    WHERE issuer_name IS NOT NULL
        AND levenshtein(issuer_name, 'PETROBRAS') <= 5
    ORDER BY distance
    LIMIT 20
""").fetchdf()

conn.close()

df
```
