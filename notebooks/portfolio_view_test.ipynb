### Country Exposures

```python
conn = duckdb.connect('../data/br_funds.db')

# Get all unique country exposures
df = conn.execute("""
    SELECT DISTINCT 
        UNNEST(string_split(country_exposure, ', ')) as country
    FROM fund_portfolio_analysis_view
    WHERE country_exposure IS NOT NULL
    ORDER BY country
""").fetchdf()

conn.close()

df
```

### assets exposure

```python
conn = duckdb.connect('../data/br_funds.db')

# Get all unique country exposures
df = conn.execute("""
    SELECT DISTINCT 
        UNNEST(string_split(asset_class_breakdown, ', ')) as asset
    FROM fund_portfolio_analysis_view
    WHERE asset_class_breakdown IS NOT NULL
    ORDER BY asset
""").fetchdf()

conn.close()

df
```
